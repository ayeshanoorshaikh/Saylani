```python
# Strings: Count the number of vowels and consonants in a string entered by the user.
user_string = input('Enter a string: ').lower()
vowels = 'aeiou'
vowel_count = 0
consonant_count = 0

for char in user_string:
    if char in vowels:
        vowel_count += 1
    elif char.isalpha():
        consonant_count += 1

print(f'Number of vowels: {vowel_count}')
print(f'Number of consonants: {consonant_count}')
```

```python
# Lists & Loops: Find the largest, smallest, and sum of list elements without using the built-in max(), min(), or sum() functions.
num=[2,3,5,1,4]
total_sum = 0
largest = num[0]
smallest = num[0]
for i in range(len(num)):
    total_sum += num[i]
    if num[i] > largest:
        largest = num[i]
    if num[i] < smallest:
        smallest = num[i]

print("Sum :", total_sum)
print("Largest:", largest)
print("Smallest:", smallest)
```

```python
# Dictionaries: Create a dictionary containing five students and their marks. Display the student who obtained the highest marks.
students_marks = {
    'Alice': 85,
    'Bob': 92,
    'Charlie': 78,
    'David': 90,
    'Eva': 88
}
highest_marks=0
top_student = ''
for student, marks in students_marks.items():
    if marks > highest_marks:
        highest_marks = marks
        top_student = student
print(f"The student with the highest marks is {top_student} with {highest_marks} marks.")
```

```python
# Functions & Loops: Create a function to check whether a number is prime. Use it to print all prime numbers between 1 and 100
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, num):
        if num % i == 0:
            return False
    return True

print("Prime numbers between 1 and 100:")
for num in range(1, 101):
    if is_prime(num):
        print(num, end=" ")
```

```python
# Final Project — Student Management System: Develop a menu-driven program using functions, loops, conditional statements, lists, and dictionaries. The program should allow users to add students, search for students, enter marks, calculate grades, and display all student records


student_records = {}

def add_student():
    name = input("Enter student name: ")
    marks = int(input("Enter student marks: "))

    if 0 <= marks <= 100:
        student_records[name] = marks
        print(f"Student {name} added successfully.")
    else:
        print("Marks must be between 0 and 100.")


def search_student():
    name = input("Enter student name to search: ")

    if name in student_records:
        print(f"{name} has {student_records[name]} marks.")
    else:
        print(f"Student {name} not found.")


def calculate_grade(marks):
    if marks >= 90:
        return 'A'
    elif marks >= 80:
        return 'B'
    elif marks >= 70:
        return 'C'
    elif marks >= 60:
        return 'D'
    else:
        return 'F'


def enter_marks():
    name = input("Enter student name to enter marks: ")

    if name in student_records:
        marks = int(input("Enter the marks: "))

        if 0 <= marks <= 100:
            student_records[name] = marks
            print("Successfully entered marks!")
        else:
            print("Marks must be between 0 and 100.")
    else:
        print("Student not found.")


def show_grade():
    name = input("Enter student name to calculate grade: ")

    if name in student_records:
        marks = student_records[name]
        grade = calculate_grade(marks)
        print(f"{name} has grade {grade}.")
    else:
        print("Student not found.")


def display_records():
    if student_records:
        for name, marks in student_records.items():
            grade = calculate_grade(marks)
            print(f"Name: {name}, Marks: {marks}, Grade: {grade}")
    else:
        print("No record found.")


while True:
    print("\nWelcome To Student Management System!")
    print("-------------------------------------")
    print("Menu:")
    print("1. Add Student")
    print("2. Search Student")
    print("3. Enter Marks of Student")
    print("4. Calculate Grade of Student")
    print("5. Display All Student Records")
    print("6. Exit\n")

    option = int(input("Select any option from Menu (1-6): "))

    if option == 1:
        add_student()

    elif option == 2:
        search_student()

    elif option == 3:
        enter_marks()

    elif option == 4:
        show_grade()

    elif option == 5:
        display_records()

    elif option == 6:
        print("Exiting Student Management System. Goodbye!")
        break

    else:
        print("Invalid Option. Please select between 1 and 6.")



        
    
```
